# Recall 제약 모델 탐색

안쪽, 바깥 Scaffold 5-Fold. 독성 Recall ≥ 0.90 조건에서 비독성 F1 최대화. 기본 4개 후보로 먼저 비교하고 새 OUTPUT 경로에서 12개 이상으로 확대한다. 바깥 Recall은 보장되지 않으며, 배포용 모델 재학습은 포함하지 않는다.


```python
from pathlib import Path
import subprocess, sys, json
import pandas as pd

TRIALS = 4
RECALL_TARGET = 0.90
OUTPUT = Path('Data/training/recall_search_pilot')
RESUME = False
command = [sys.executable, '-m', 'analysis.search_recall', '--trials', str(TRIALS),
           '--recall-target', str(RECALL_TARGET), '--output', str(OUTPUT)]
subprocess.run(command + ['--dry-run'], check=True)
```

```python
# 기존 결과를 덮어쓰지 않는다. 중단된 동일 실행을 재개할 때만 RESUME=True.
subprocess.run(command + (['--resume'] if RESUME else []), check=True)
```

```python
display(pd.read_csv(OUTPUT / 'outer_metrics.csv'))
display(pd.read_csv(OUTPUT / 'baseline_comparison.csv'))
print(json.loads((OUTPUT / 'evaluation.json').read_text()))
```
