# 유망한 모델의 피처 선택 비교

LightGBM, XGBoost의 Fold별 기존 안쪽 최고 설정을 고정한다. 선택 없음, ANOVA, gain 중요도와 Top-K를 비교한다. 독성 Recall ≥ 0.87 조건에서 비독성 F1 최대화. 같은 분할의 선택 없는 대조군을 함께 평가하며, 바깥 Recall은 보장되지 않는다.


```python
from pathlib import Path
import subprocess, json
import pandas as pd

PYTHON = Path('.venv/bin/python').resolve()
assert PYTHON.is_file()
SOURCE = Path('Data/training/optuna_recall87_v1')
OUTPUT = Path('Data/training/focused_features_recall87_v1')
RESUME = True
command = [str(PYTHON), '-u', '-m', 'analysis.focused_features',
           '--source', str(SOURCE), '--output', str(OUTPUT), '--recall-target', '0.87']
subprocess.run(command + ['--dry-run'], check=True)
```

```python
# 같은 출력 폴더에서 다른 프로세스가 실행 중이면 동시에 실행하지 않는다.
subprocess.run(command + (['--resume'] if RESUME else []), check=True)
```

```python
display(pd.read_csv(OUTPUT / 'comparison.csv'))
display(pd.read_csv(OUTPUT / 'outer_metrics.csv'))
print(json.loads((OUTPUT / 'evaluation.json').read_text()))
```
